## Project Description:
## HomeGuard is a simulated smart home alarm system.
## It monitors motion, door, temperature, and smoke sensors.
## The system generates simulated sensor readings and checks them
## based on whether the house is in HOME or AWAY mode.
## If an abnormal or dangerous event is detected, the system
## generates and logs an appropriate alert.

```python
import random
import datetime


class HomeSecurity:

    def __init__(
        self,
        mode="Away",
        motion_sensor=False,
        door="Closed",
        temp_sensor=None,
        smoke_sensor=False
    ):
        self.mode = mode
        self.motion_sensor = motion_sensor
        self.door = door

        if temp_sensor is None:
            self.temp_sensor = random.randint(30, 100)
        else:
            self.temp_sensor = temp_sensor

        self.smoke_sensor = smoke_sensor
        self.security_status = "Normal"
        self.alert_source = None


    def update_mode(self):
        """Check sensors and update the security status."""

        self.security_status = "Normal"
        self.alert_source = None

        if self.smoke_sensor:
            self.security_status = "Critical"
            self.alert_source = "Smoke"

        elif self.temp_sensor < 35 or self.temp_sensor > 95:
            self.security_status = "Safety"
            self.alert_source = "Temperature"

        # AWAY mode
        elif self.mode == "Away":

            if self.motion_sensor:
                self.security_status = "High Security"
                self.alert_source = "Motion"

            elif self.door == "Open":
                self.security_status = "High Security"
                self.alert_source = "Door"

        # HOME mode
        elif self.mode == "Home":

            # Comfort notice when temperature is outside 65–75°F
            if self.temp_sensor < 65 or self.temp_sensor > 75:
                self.security_status = "Comfort"
                self.alert_source = "Temperature"


    def display_status(self):
        """Display readings and alerts in the required format."""

        self.update_mode()

        current_time = datetime.datetime.now().strftime("%H:%M:%S")

        print()
        print("=== HomeGuard Security System ===")
        print("Time:", current_time)
        print("Mode:", self.mode.upper())
        print()

        # Motion
        if self.motion_sensor:
            print("[READING] Living Room Motion: MOTION DETECTED")
        else:
            print("[READING] Living Room Motion: No activity")

        # Door
        print("[READING] Front Door:", self.door.upper())

        # Temperature
        if 65 <= self.temp_sensor <= 75:
            temp_status = "Normal"
        else:
            temp_status = "Abnormal"

        print(
            f"[READING] Kitchen Temperature: "
            f"{self.temp_sensor}°F ({temp_status})"
        )

        # Smoke
        if self.smoke_sensor:
            print("[READING] Bedroom Smoke: SMOKE DETECTED")
        else:
            print("[READING] Bedroom Smoke: CLEAR")

        print()

        # Alerts
        if self.security_status == "Critical":
            print(
                "[ALERT!] 🔥 CRITICAL: SAFETY: "
                "Smoke detected in Bedroom!"
            )

        elif self.security_status == "Safety":
            print(
                f"[ALERT!] ⚠️ HIGH: SAFETY: "
                f"Kitchen temperature is {self.temp_sensor}°F!"
            )

        elif self.security_status == "High Security":

            if self.alert_source == "Door":
                print(
                    f"[ALERT!] 🚨 HIGH: SECURITY: "
                    f"Front Door opened while in {self.mode.upper()} mode!"
                )

            elif self.alert_source == "Motion":
                print(
                    f"[ALERT!] 🚨 HIGH: SECURITY: "
                    f"Motion detected while in {self.mode.upper()} mode!"
                )

        elif self.security_status == "Comfort":
            print(
                f"[ALERT!] 💡 LOW: COMFORT: "
                f"Kitchen temperature is {self.temp_sensor}°F."
            )

        # Log only if something abnormal happened
        if self.security_status != "Normal":
            print(
                f"[LOG] [{current_time}] "
                "Sending notification to homeowner..."
            )
    
```

```python
print("HOME test")
home1 = HomeSecurity(mode="HOME", motion_sensor=False, door="Closed", smoke_sensor=False)
home1.display_status()
print("AWAY Test")
home2 = HomeSecurity(mode="AWAY", motion_sensor=True, door="Open", temp_sensor=50, smoke_sensor=True)
home2.display_status()
print("SLEEP Test")
home3 = HomeSecurity(mode="SLEEP", motion_sensor=False, door="Open", temp_sensor=45, smoke_sensor=False)
home3.display_status()
```
